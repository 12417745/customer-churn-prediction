## Day 4.1 — Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
decision_tree = DecisionTreeClassifier(random_state=42)
```

```python
tree_precision = precision_score(y_test, y_pred_tree)
tree_recall = recall_score(y_test, y_pred_tree)
tree_f1 = f1_score(y_test, y_pred_tree)

print("Decision Tree Precision: {:.2f}%".format(tree_precision * 100))
print("Decision Tree Recall: {:.2f}%".format(tree_recall * 100))
print("Decision Tree F1-score: {:.2f}%".format(tree_f1 * 100))
```

## 4.2 Random Forest

Random Forest is an ensemble machine learning algorithm that combines multiple decision trees to make predictions.

from sklearn.ensemble import RandomForestClassifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
random_forest.fit(X_train_scaled, y_train)
```

```python
y_pred_rf = random_forest.predict(X_test_scaled)
```

```python
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", rf_accuracy)
print("Random Forest Accuracy: {:.2f}%".format(rf_accuracy * 100))
```

```python
rf_precision = precision_score(y_test, y_pred_rf)
rf_recall = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)

print("Random Forest Precision: {:.2f}%".format(rf_precision * 100))
print("Random Forest Recall: {:.2f}%".format(rf_recall * 100))
print("Random Forest F1-score: {:.2f}%".format(rf_f1 * 100))
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy,
        tree_accuracy,
        rf_accuracy
    ],
    "Precision": [
        precision,
        tree_precision,
        rf_precision
    ],
    "Recall": [
        recall,
        tree_recall,
        rf_recall
    ],
    "F1-Score": [
        f1,
        tree_f1,
        rf_f1
    ]
})

comparison
```

```python
comparison_display = comparison.copy()

for column in ["Accuracy", "Precision", "Recall", "F1-Score"]:
    comparison_display[column] = (
        comparison_display[column] * 100
    ).round(2)

comparison_display
```

```python
import matplotlib.pyplot as plt

comparison_display.set_index("Model")[[
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Model Performance Comparison")
plt.ylabel("Score (%)")
plt.xlabel("Model")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

## 4.3 Model Performance Comparison

The three machine learning models were evaluated using accuracy, precision, recall, and F1-score.

The results show that the models have different performance across the evaluation metrics. Logistic Regression achieved 82.04% accuracy, while Decision Tree achieved 71.19% and Random Forest achieved 78.71%.

Precision, recall, and F1-score are also considered because customer churn prediction involves identifying customers who may leave.
